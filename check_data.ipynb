```python
# parse the parquet back to the csv file, for for-desktop to use for TC-CLiFF-LHMP


import pandas as pd
import os

# Load the parquet file
df = pd.read_parquet('dataset/atc/full_10/atc1_test_split.parquet')

# Extract the first 4 digits of person_id and group by them
df['group_key'] = df['person_id'].astype(str).str[:4]

# Strip first 4 digits and remove trailing .0
df['person_id'] = df['person_id'].astype(str).str[4:].str.replace(r'\.0$', '', regex=True).astype(int)

output_dir = '/home/yufei/research/for-desktop/atc-1s-ds-eccv/'
os.makedirs(output_dir, exist_ok=True)

for key, group in df.groupby('group_key'):
    output_path = f'{output_dir}{key}.csv'
    group.drop(columns='group_key').to_csv(output_path, index=False, header=False)
    print(f'Saved {output_path} with {len(group)} rows')
```
